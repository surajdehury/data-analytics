```python
import pandas as pd
data=pd.read_csv('data_clean.csv')
```

```python
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
data=data.drop('Unnamed: 0',axis=1)
```

```python
data=data.drop('Temp C',axis=1)
```

```python
data.head(3)
```

```python
data=data.rename({'Solar.R':'Solar'},axis=1)
```

```python
data.head()
```

```python
data[data.duplicated()].shape
```

```python
data[data.duplicated()]
```

```python
data=data.drop_duplicates()# dropped the duplicated data from list
```

```python
data['Month'].isnull().sum()
```

```python
data['Month'] = data['Month'].replace({
    'January':1, 'February':2, 'March':3,
    'April':4, 'May':5, 'June':6,
    'July':7, 'August':8, 'September':9,
    'October':10, 'November':11, 'December':12
})
```

```python
data['Month'].head(50)
```

```python
data['Weather'].isnull().sum()
```

```python
#conversion of numeric to category example
data['Weather']=data['Weather'].astype('category')
data['Temp']=pd.to_numeric(data['Temp'],errors='coerce')
```

```python
data.isnull().sum()
```

```python
m=data['Ozone'].mean()
data['Ozone']=data['Ozone'].fillna(m)
data.isnull().sum()
```

```python
n=data['Solar'].mean()
data['Solar']=data['Solar'].fillna(n)
data.isnull().sum()
```

```python
o=data['Weather'].mode().iloc[0]
data['Weather']=data['Weather'].fillna(o)
data.isnull().sum()
```
