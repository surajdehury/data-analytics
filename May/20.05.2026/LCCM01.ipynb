```python
import pandas as pd
```

```python
sal_data=pd.read_csv('sal_data.csv')
bonus_data=pd.read_csv('bonus_data.csv')
```

```python
#using merge display common employees in both files
```

```python
outer_join=pd.merge(sal_data,bonus_data,how='outer')
print(outer_join)
```

```python
#common ele present in 1st csv and 2nd csv
```

```python
inner_join=pd.merge(sal_data,bonus_data,how='inner')
print(inner_join)
```

```python
#all datas of sal data and common data of bonus data
```

```python
left_join=pd.merge(sal_data,bonus_data,how='left')
print(left_join)
```

```python
#replace nan with best possible value
```

```python
outer_join=pd.merge(sal_data,bonus_data,how='outer')
n=outer_join['Bonus'].mean()
outer_join['Bonus']=outer_join['Bonus'].fillna(n)
print(outer_join['Bonus'])
```
