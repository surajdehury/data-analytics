```python
import pandas as pd
sal_data=pd.read_csv('sal_data.csv')
bonus_data=pd.read_csv('bonus_data.csv')
#using merge display common employees in both files
outer_join=pd.merge(sal_data,bonus_data,how='outer')
print(outer_join)
```

```python
#common data which is present in first and second data
inner_join=pd.merge(sal_data,bonus_data,how='inner')
print(inner_join)
```

```python
#displpay all the columns of the employees of sal data and matching ele of 2nd data
left_join=pd.merge(sal_data,bonus_data,how='left')
print(left_join)
```

```python
#replace nan with best possible values
n=outer_join['Bonus'].mean()
outer_join['Bonus']=outer_join['Bonus'].fillna(n)
print(outer_join['Bonus'])
```

```python
#use for loops for time h sum of all time
sum=0
for t in data['TIME (h)']:
    sum+=t
print(sum)
```

```python
#csv file 
#basic_salary 2
data2=pd.read_csv('basic_salary_2.csv')
data2.head(5)
```

```python
#display columnwise na values
data2.isnull().sum()
```

```python
#fill the nan with median
o=data2['ms'].median()
data2['ms']=data2['ms'].fillna(o)
print(o)
print(data2['ms'])
```

```python
#display how many different values in location column
data2['Location'].value_counts()
```

```python
#find out the outliers in each column of numeric type
import matplotlib.pyplot as plt
data_ba=data2.ba
plt.boxplot(data_ba)
```

```python
import matplotlib.pyplot as plt
data_ms=data2.ms
plt.boxplot(data_ms)
```

```python
#display the dataset value where location is mumbai
A=data2[data2['Location']=='MUMBAI']
print(A)
```

```python
#display the datatset where ba is greater than 1300
B=data2[data2['ba']>13000]
print(B)
```

```python
#display min,max,and mean value of the ba
min=data2['ba'].agg('min')
print(min)
max=data2['ba'].agg('max')
print(max)
mean=data2['ba'].agg('mean')
print(mean)
```

```python
#mode of the column grade
p=data2['Grade'].mode()
data2['Grade']=data2['Grade'].fillna(p)
print(p)
```
